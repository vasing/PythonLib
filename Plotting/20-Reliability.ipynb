# Exponential Distribution

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range of t values
t = np.linspace(0, 5, 100)  # 100 points from 0 to 2

# Calculate the exponential function
Lambda1 = 0.5 # scale parameter
Lambda2 = 1   # scale parameter
Lambda3 = 1.5 # scale parameter
Gamma = 0     # location parameter

y1 = Lambda1 * np.exp(-Lambda1 * t - Gamma)
y2 = Lambda2 * np.exp(-Lambda2 * t - Gamma)
y3 = Lambda3 * np.exp(-Lambda3 * t - Gamma)

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(t, y1, label='Lambda = 0.5', color='blue')
plt.plot(t, y2, label='Lambda = 1', color='green')
plt.plot(t, y3, label='Lambda = 1.5', color='red')
plt.title('Exponential Destribution variant Lambda')
plt.xlabel('t')
plt.ylabel('f(t)')
plt.axhline(0, color='black',linewidth=0.5, ls='--')
plt.axvline(0, color='black',linewidth=0.5, ls='--')
plt.grid()
plt.legend()
plt.show()
```

# Weibull Distribution

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters
beta = 2.0   # Example value for beta
eta = 1.0    # Example value for eta
gamma = 0.0  # Example value for gamma

# Define the function f(t)
def f(t):
    term = (t - gamma) / eta
    return (beta / eta) * (term ** (beta - 1)) * np.exp(-term ** beta)

# Define the range of t values
t_values = np.linspace(0, 3, 100)  # 100 points from -2 to 5

# Calculate f(t) for the defined t values
f_values = f(t_values)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(t_values, f_values, label='Beta = 2, Eta = 1, Gamma = 0', color='blue')
plt.title('Weibull Distribution')
plt.xlabel('t')
plt.ylabel('f(t)')
plt.axhline(0, color='black', linewidth=0.5, ls='--')
plt.axvline(0, color='black', linewidth=0.5, ls='--')
plt.grid()
plt.legend()
plt.ylim(bottom=0)  # Set the lower limit of y-axis to 0 for better visibility
plt.show()
```
