```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range of t values
t = np.linspace(0, 5, 100)  # 100 points from 0 to 2

# Calculate the exponential function
Lambda1 = 0.5 # scale parameter
Lambda2 = 1   # scale parameter
Lambda3 = 1.5   # scale parameter

y1 = Lambda1 * np.exp(-Lambda1 * t)
y2 = Lambda2 * np.exp(-Lambda2 * t)
y3 = Lambda3 * np.exp(-Lambda3 * t)

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(t, y1, label='Lambda = 0.5', color='blue')
plt.plot(t, y2, label='Lambda = 1', color='green')
plt.plot(t, y3, label='Lambda = 1.5', color='red')
plt.title('Exponential Destribution Plot')
plt.xlabel('t')
plt.ylabel('f(t)')
plt.axhline(0, color='black',linewidth=0.5, ls='--')
plt.axvline(0, color='black',linewidth=0.5, ls='--')
plt.grid()
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the range of t values
t = np.linspace(0, 5, 100)  # 100 points from 0 to 2

# Calculate the exponential function
Lambda = 1 # scale parameter
Gamma1 = 0    # location parameter
Gamma2 = 0.5  # location parameter
Gamma3 = 1    # location parameter

print(Lambda)

y1 = Lambda * np.exp(-Lambda * (t - Gamma1))
y2 = Lambda * np.exp(-Lambda * (t - Gamma2))
y3 = Lambda * np.exp(-Lambda * (t - Gamma3))

# Create the plot
plt.figure(figsize=(8, 5))
plt.plot(t, y1, label='Gamma = 0', color='blue')
plt.plot(t, y2, label='Gamma = 0.5', color='green')
plt.plot(t, y3, label='Gamma = 1', color='red')
plt.title('Exponential Function Plot')
plt.xlabel('x')
plt.ylabel('f(t)')
plt.axhline(0, color='black',linewidth=0.5, ls='--')
plt.axvline(0, color='black',linewidth=0.5, ls='--')
plt.grid()
plt.legend()
plt.show()
```
