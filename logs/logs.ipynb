# 1. Simple Terminal Script (Linux/macOS)

```python
#!/usr/bin/env python3
# /home/username/scripts/terminal_test.py

import os
import sys
import datetime
import subprocess

def main():
    print(f"Script executed at: {datetime.datetime.now()}")
    print(f"Python path: {sys.executable}")
    print(f"Working directory: {os.getcwd()}")
    print(f"User: {os.getenv('USER', 'unknown')}")
    print(f"Home: {os.getenv('HOME', 'unknown')}")
    
    # Create a temporary file with our output
    timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    log_file = f"/tmp/cron_test_{timestamp}.txt"
    
    with open(log_file, 'w') as f:
        f.write(f"Cron Test Execution\n")
        f.write(f"Time: {datetime.datetime.now()}\n")
        f.write(f"Python: {sys.executable}\n")
        f.write(f"Version: {sys.version}\n")
        f.write(f"PID: {os.getpid()}\n")
        f.write(f"User: {os.getenv('USER')}\n")
        f.write("="*50 + "\n")
        f.write("This proves the script ran successfully!\n")
    
    print(f"Log written to: {log_file}")
    print("Test completed successfully!")
    
    # OPTIONAL: Actually open a terminal to show the message
    # Uncomment the lines below if you want to see a popup terminal
    
    # message = f"Script ran at: {datetime.datetime.now()}\nPython: {sys.executable}"
    # subprocess.run(['xterm', '-hold', '-e', f'echo "{message}" && echo "Press Enter to close..." && read'])
    # OR for gnome-terminal:
    # subprocess.run(['gnome-terminal', '--', 'bash', '-c', f'echo "{message}"; echo "Press Enter..."; read'])

if __name__ == "__main__":
    main()
```

# 2. Script with GUI Notification (Better for Cron)

```python
#!/usr/bin/env python3
# /home/username/scripts/cron_notify.py

import os
import sys
import datetime
import subprocess
import json

def main():
    # Create detailed output
    output = [
        "=" * 60,
        f"CRON JOB TEST - {datetime.datetime.now()}",
        f"Python Executable: {sys.executable}",
        f"Python Version: {sys.version.split()[0]}",
        f"Working Directory: {os.getcwd()}",
        f"User: {os.getenv('USER', 'unknown')}",
        f"Home Directory: {os.getenv('HOME', 'unknown')}",
        f"PATH: {os.getenv('PATH', 'unknown')[:100]}...",
        f"Process ID: {os.getpid()}",
        f"Parent Process ID: {os.getppid()}",
        "=" * 60,
        "✓ Script executed successfully!",
        "✓ This proves cron is working correctly.",
        "=" * 60
    ]
    
    # Print to stdout (will be captured by cron logs)
    for line in output:
        print(line)
    
    # Write to a log file
    log_dir = "/tmp/cron_logs"
    os.makedirs(log_dir, exist_ok=True)
    
    log_file = os.path.join(log_dir, f"cron_test_{datetime.datetime.now().strftime('%Y%m%d_%H%M%S')}.log")
    with open(log_file, 'w') as f:
        f.write('\n'.join(output))
    
    print(f"\nLog saved to: {log_file}")
    
    # Try to send a desktop notification (if DISPLAY is available)
    try:
        notify_cmd = [
            'notify-send',
            'Cron Job Test',
            f'Script executed successfully at {datetime.datetime.now().strftime("%H:%M:%S")}\nSee {log_file} for details',
            '-t', '5000'  # Display for 5 seconds
        ]
        subprocess.run(notify_cmd, check=False)
    except:
        pass  # Notification not available, that's OK
    
    # Also create a simple popup using zenity if available
    try:
        popup_cmd = [
            'zenity', '--info',
            '--title', 'Cron Test Result',
            '--text', f'Cron job executed!\nTime: {datetime.datetime.now()}\nPython: {sys.executable}\nLog: {log_file}',
            '--width', '400',
            '--height', '200'
        ]
        subprocess.run(popup_cmd, check=False)
    except:
        pass  # zenity not available

if __name__ == "__main__":
    main()
```
