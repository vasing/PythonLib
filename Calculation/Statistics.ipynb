# Statistik

Absolute Häufigkeit einer Stichprobe berechnen

Häufigkeitsfunktion in einem Stabdigramm darstellen 

```python
import pandas as pd
import matplotlib.pyplot as plt

# Stichprobe
daten = [97, 98, 98, 98, 99, 99, 99, 99, 100, 100, 100, 100, 100, 100, 100, 100, 100, 101, 101, 101, 101, 101, 102, 102, 103]

# Absolute Häufigkeit
haeufigkeit = pd.Series(daten).value_counts()

# Daten anzeigen
print("Absolute Häufigkeit der Stichprobe:")
print(haeufigkeit)

# Stabdiagramm (Balkendiagramm) erstellen
plt.bar(haeufigkeit.index, haeufigkeit.values, width=0.4, color='skyblue', edgecolor='black')
plt.title('Absolute Häufigkeit der Stichprobe')
plt.xlabel('Werte')
plt.ylabel('f(x)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Stichprobe
# 500 mal Wurf eines Würfels
# 1: 102 mal, 2: 88 mal, 3: 80 mal, 4: 90 mal, 5: 65 mal, 6: 75 mal
daten = [1] * 102 + [2] * 88 + [3] * 80 + [4] * 90 + [5] * 65 + [6] * 75

# Absolute Häufigkeit
haeufigkeit = pd.Series(daten).value_counts().sort_index()

# Relative Häufigkeiten und Summenhäufigkeiten (Verteilungsfunktion)
relative_haeufigkeit = haeufigkeit / haeufigkeit.sum()
verteilungsfunktion = relative_haeufigkeit.cumsum()

# Daten anzeigen
print("Absolute Häufigkeit der Stichprobe:")
print(haeufigkeit)
print("\nRelative Häufigkeiten und Summenhäufigkeiten:")
print(pd.DataFrame({
    'absolute_häufigkeit': haeufigkeit.values,
    'relative_häufigkeit': relative_haeufigkeit.values,
    'summenhäufigkeit': verteilungsfunktion.values
}, index=haeufigkeit.index))

# Treppenfunktion der Verteilungsfunktion darstellen
x = verteilungsfunktion.index.to_numpy()
y = verteilungsfunktion.to_numpy()
plt.step(x, y, where='post', color='darkgreen', linewidth=2, label='F(x)')
plt.scatter(x, y, color='darkgreen', zorder=3)
plt.title('Verteilungsfunktion der Stichprobe')
plt.xlabel('Würfelaugen')
plt.ylabel('F(x) = P(X <= x)')
plt.xticks(x)
plt.ylim(0, 1.05)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```
