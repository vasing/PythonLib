# Complex Numbers

## Get Complex Number and Parts

```python
import math

z = 2 + 3j
z = complex(2, 3)

print(z)
```

```python
z = 2 + 3j

print(z.real)
print(z.imag)
print(abs(z))
print(z.conjugate())
```

```python
# Use cmath (not math) for complex inputs — math.sqrt(-1) raises an error.
import cmath

cmath.sqrt(z)         # principal square root
cmath.exp(z)
cmath.log(z)          # natural log
cmath.log10(z)
cmath.sin(z); cmath.cos(z); cmath.tan(z)
cmath.sinh(z); cmath.cosh(z); cmath.tanh(z)
cmath.asin(z); cmath.acos(z); cmath.atan(z)
cmath.phase(z)        # angle in radians
cmath.polar(z)        # (r, theta) tuple
r=abs(z)
theta = cmath.phase(z) 
cmath.rect(r, theta)  # r*e^(iθ)
cmath.isnan(z); cmath.isinf(z); cmath.isfinite(z)
```

# Calculating Complex Numbers

```python
a = 3 + 4j
b = 1 - 2j

a + b       # (4+2j)
a - b       # (2+6j)
a * b       # (11-2j)
a / b       # (-1+2j)
a ** 2      # (-7+24j)
-a          # (-3-4j)
```

# Plotting Complex Numbers
## Vector in Gaußian Plane

```python
import matplotlib.pyplot as plt

def plot_complex_vectors(vectors, labels=None, title="Gaussian Number Plane"):
    """
    vectors : list of complex numbers
    labels  : optional list of strings
    """
    fig, ax = plt.subplots(figsize=(6, 6))

    # Axes through origin
    ax.axhline(0, color='black', lw=0.8)
    ax.axvline(0, color='black', lw=0.8)

    # Draw each complex number as a vector from origin
    for i, z in enumerate(vectors):
        ax.arrow(0, 0, z.real, z.imag,
                 head_width=0.15, head_length=0.2,
                 length_includes_head=True,
                 color=f"C{i}", lw=2, alpha=0.9)

        # Label at the tip
        text = labels[i] if labels else f"{z:.2f}"
        ax.text(z.real*1.05, z.imag*1.05, text,
                color=f"C{i}", fontsize=11, fontweight='bold')

    # Gaussian grid: unit squares
    ax.set_aspect('equal')
    ax.grid(True, which='both', ls=':', color='gray', alpha=0.6)
    ax.set_xticks(range(-5, 6))
    ax.set_yticks(range(-5, 6))

    # Labels on axes (Re / Im)
    ax.set_xlabel("Re (real axis)", fontsize=11)
    ax.set_ylabel("Im (imaginary axis)", fontsize=11)
    ax.set_title(title, fontsize=13)

    # Put the 'i' label near positive imaginary axis
    ax.text(0.15, max(v.imag for v in vectors) + 0.3, "i",
            fontsize=12, fontstyle='italic')

    plt.tight_layout()
    plt.show()


# Example
zs = [3+4j, -2+1j, 1-3j, 2+2j, -1-1j]
plot_complex_vectors(zs, labels=["z₁","z₂","z₃","z₄","z₅"])
```

```python
import matplotlib.pyplot as plt
import numpy as np

z = np.array([3+4j, -2+1j, 1-3j, 2+2j, -1-1j])
fig, ax = plt.subplots()
ax.axhline(0, color='k', lw=0.5); ax.axvline(0, color='k', lw=0.5)
ax.scatter(z.real, z.imag, c='crimson', zorder=3)
for zi in z:
    ax.annotate(f"{zi:.1f}", (zi.real, zi.imag), xytext=(5,5),
                textcoords='offset points')
ax.set_aspect('equal'); ax.grid(True, ls=':')
ax.set(xlabel='Re', ylabel='Im', title='Argand Diagram')
plt.show()
```

## Vector Arrows from Origin

```python
fig, ax = plt.subplots()
for zi in z:
    ax.arrow(0, 0, zi.real, zi.imag, head_width=0.15,
             length_includes_head=True, color='steelblue')
ax.set_aspect('equal'); ax.grid(True)
plt.show()
```

## Polar Plot

```python
fig = plt.figure()
ax = fig.add_subplot(projection='polar')
ax.scatter(np.angle(z), np.abs(z), c='purple')
ax.set_title('Complex numbers in polar form')
plt.show()
```
